```python
import pandas as pd
import numpy as np
from sklearn.svm import SVR
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Salary_Data.csv")

data.head()
```

```python
data.info()
```

```python
X = data['YearsExperience']
y = data['Salary']

X = np.array(X)
X = X[:,np.newaxis]
```

```python
model = SVR(C=1000, gamma=0.05, kernel='rbf')

model.fit(X, y)
```

```python
plt.scatter(X, y)
plt.plot(X, model.predict(X))
```
